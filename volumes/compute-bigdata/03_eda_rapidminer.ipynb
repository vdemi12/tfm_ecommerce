## EDA de ratings

El análisis K-Means con k=4 sobre las dimensiones RFM (Recencia, Frecuencia y Rating medio) identificó cuatro segmentos diferenciados de usuarios. El segmento mayoritario, cluster_0 (n≈17.000), agrupa usuarios ocasionales con frecuencia de interacción por debajo de la media y comportamiento neutro en recencia y valoración. El cluster_1 (n≈2.000) concentra usuarios activos con frecuencia significativamente superior a la media y actividad reciente, constituyendo el segmento prioritario para el motor de recomendación. El cluster_3 (n≈96) representa usuarios hiper-activos con una frecuencia 8,6 desviaciones típicas por encima de la media, posiblemente revisores profesionales o usuarios con comportamiento automatizado. Finalmente, el cluster_2 (n=29) agrupa usuarios atípicos con timestamps extremadamente antiguos y valoraciones consistentemente bajas, por lo que este segmento fue excluido del motor de recomendación al no representar comportamiento de compra normalizado.

## EDA de fraud_processed

El análisis se realizó mediante un árbol de decisión (criterio: information gain) entrenado sobre una muestra estratificada de 50.000 transacciones, con partición 70/30 para entrenamiento y test respectivamente. Las variables TransactionID y card1 se asignaron como identificadores, isFraud como variable objetivo, y se excluyeron las features con correlación superior a 0.97 identificadas en la matriz de correlación previa (C2, C6, C11, D2).
El modelo alcanzó una accuracy del 97.89% en entrenamiento y 97.01% en test. Sin embargo, dado el severo desbalance de clases (tasa de fraude del 3.5%), la accuracy no es una métrica representativa del rendimiento real. El recall sobre la clase positiva (fraude) fue del 42.62% en entrenamiento y 32.50% en test, evidenciando que el modelo clasifica correctamente menos de un tercio de los fraudes reales en datos no vistos. La precisión sobre la clase positiva fue del 91.62% en entrenamiento y 62.92% en test.
Estos resultados confirman empíricamente la necesidad de técnicas de balanceo de clases para abordar este problema. En consecuencia, la fase de modelado implementará SMOTE sobre el conjunto de entrenamiento y utilizará una arquitectura LSTM con TensorFlow/Keras, optimizando las métricas F1 y Recall en lugar de la accuracy, tal como se justifica en este análisis exploratorio.
